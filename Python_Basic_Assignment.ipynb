#Assignment 1 Python Basics

1. What is Python, and why is it popular?
  -  Python is a high-level, easy to understand programming language. This language is popular because of its simplicity, versatility, large community, extensive libraries, and due to the use of this language in web development to machine learning.

2. What is an interpreter in Python?
  -  Interpreter is basically a program that reads and executes code line by line. Before the process of execution, it translates a high level Python code into a machine readable bytecode.

3. What are pre-defined keywords in Python?
  -  In python, some keywords are pre-defined, or reserved with special meanings so the interpreter can understand them. For example,  
True, False, None, else, return, for, and, etc,.

4. Can keywords be used as variable names?
  -  As we already know that Keywords are reserved words that have predefined meanings in Python. So, we cannot use these keywords as variable names in Python. If we use a keyword as a variable name, it will give a syntax error.

5. What is mutability in Python?
  - Mutability refers the object in which content can be modified without creating a new object In Python. For example, tuples are immutable because we cannot modify them after creation.

6. Why are lists mutable, but tuples are immutable?
  -  Lists are mutable because we can modify them after creation. For example, x = (1, 3, 5, 6, 7) is a list and we can easily modify this list by using append(). On the other hand, tuples are immutable because we cannot modify them after creation.

7. What is the difference between “==” and “is” operators in Python?
  -  The operators, “==” (equality operator) and “is” (identity operator) both are used in Python as comparison operators, but, in different ways. For example, operator, “==” checks if two objects have the same value. On the other hand, operator “is” checks if two variables point to the same object in memory.

8. What are logical operators in Python?
  -  The logical operators also known as Boolean operators in Python are and, or, and not. These operators allow us to combine and evaluate conditions, resulting in either True or False.

9. What is type casting in Python?
  -  In Python, the type casting is the process to change the data type of a variable. For example, we can convert a float or string to an integer using int(). Also, we can convert an integer or string to a float using float().

10. What is the difference between implicit and explicit type casting?
  -  Implicit type conversion or Explicit type conversion, both are the examples of type casting. But, Implicit type conversion during a programming is handled automatically by the compiler on the other hand, Explicit type conversion is done manually by the programmer  in programming.

11. What is the purpose of conditional statements in Python?
  -  If we want to control the flow of a program by executing different blocks of code based on conditions of True or False, we use conditional statements. For this, we can us ‘if’, ‘else’ and ‘elif’.

12. How does the elif statement work?
  -  In Python, the “elif” keyword stands for “else if”. We can use this keyword in conditional statements to check for multiple conditions. For example, if the first condition is false, it moves on to the next “elif” statement to check if that condition is true.

13. What is the difference between for and while loops?
  -  In python, a for loop repeats the execution of a code block from a sequence till certain number of times. On the other hand, while loop repeats the execution of a code block till the condition is true, means number of executions not certain here.

14. Describe a scenario where a while loop is more suitable than a for loop.
  -  A while loop is more suitable than a for loop when the number of iterations is not certain and depends on a condition. For example, we are writing codes for a program in which a user must enter a password to proceed. The program keeps prompting the user until the user provide the correct password. Now, since, we don't know the exact number of attempts by the user to reach a correct password, a while loop is a good choice here.

```python
#Q10. Write code to demonstrate type casting with list elements.
Pawan_list = ["10", "20", "30", "40", "50"]

int_list = [int(num) for num in Pawan_list]
print("List with integers:", int_list)

float_list = [float(num) for num in int_list]
print("List with floats:", float_list)

string_list = [str(num) for num in float_list]
print("List with strings:", string_list)

set_example = set(int_list)
print("List converted to set:", set_example)
```

```python
#Q14. Write a program to reverse a string using a while loop.
original_string = input("Enter a string: ")
reversed_string = ""
index = len(original_string) - 1
while index >= 0:
    reversed_string += original_string[index]
    index -= 1
print(f"Reversed string: {reversed_string}")
```

```python
# Q15. Write a Python program to calculate the factorial of a number provided by
# the user using a while loop.
# Taking user input
num = int(input("Enter a number: "))
factorial = 1
i = num
while i > 0:
    factorial *= i
    i -= 1
print(f"The factorial of {num} is: {factorial}")
```
